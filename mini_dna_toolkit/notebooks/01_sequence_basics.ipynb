##### This notebook explores basic operations that are commonly used in DNA sequence analysis. 

```python
seq = "ATGCGTACGTTAGCTAGCTAGGCTA"
print(seq)
```

```python
len(seq)
```

```python
a_count = seq.count("A")
t_count = seq.count("T")
g_count = seq.count("G")
c_count = seq.count("C")

print("A:", a_count)
print("T:", t_count)
print("G:", g_count)
print("C:", c_count)
```

```python
total_bases = a_count + t_count + g_count + c_count

print("sequence length:", len(seq))
print("total bases:", total_bases)
```

```python
a_percentage = a_count / len(seq) * 100
t_percentage = t_count / len(seq) * 100
g_percentage = g_count / len(seq) * 100
c_percentage = c_count / len(seq) * 100

print("A:", round(a_percentage, 2), "%")
print("T:", round(t_percentage, 2), "%")
print("G:", round(g_percentage, 2), "%")
print("C:", round(c_percentage, 2), "%")
```

```python
total_percentage = (a_percentage + t_percentage + g_percentage + c_percentage)
print(round(total_percentage, 2))
```

```python
gc_count = g_count + c_count
gc_percentage = gc_count / len(seq) * 100

print("GC count:", gc_count)
print("GC percentage:", round(gc_percentage, 2), "%")
```

```python
at_count = a_count + t_count
at_percentage = at_count / len(seq) * 100

print("AT percentage:", round(at_percentage, 2), "%")
print("GC percentage:", round(gc_percentage, 2), "%")
```

```python
print("AT + GC:", round(at_percentage + gc_percentage, 2), "%")
```

```python
if gc_percentage < 40:
    print("the sequence has low GC content.")
elif gc_percentage > 60:
    print("the sequence has high GC content.")
else:
    print("the sequence has moderate GC content.")
```

```python
complement_map = {
    "A": "T",
    "T": "A",
    "C": "G",
    "G": "C"
}
```

```python
seq = "ATCAGTCC"
complement = ""
for base in seq:
    complement += complement_map[base]
print("sequence:", seq)
print("complement sequence:", complement)
```

```python
seq = "ATGCCAT"
complement = ""
for base in seq:
    complement += complement_map[base]
reverse_complement = complement[::-1]

print("sequence:", seq)
print("complement:", complement)
print("reverse complement:", reverse_complement)
```

```python
dna_seq = "ATGCCATG"
rna_seq = dna_seq.replace("T", "U")

print("DNA:", dna_seq)
print("RNA:", rna_seq)
```

```python
codon_table = {
    "AUG": "M",
    "GCC": "A",
    "GCU": "A",
    "GCA": "A",
    "GCG": "A",
    "UAA": "*", # * means stop codon
    "UAG": "*",
    "UGA": "*"
}
```

```python
rna_seq = "AUGGCCAUG"
for i in range(0, len(rna_seq), 3):
    codon = rna_seq[i:i + 3]
    print(codon)
```

```python
protein = ""
for i in range(0, len(rna_seq), 3):
    codon = rna_seq[i:i + 3]

    if len(codon) < 3:
        break

    amino_acid = codon_table.get(codon)

    if amino_acid == "*":
        break

    protein += amino_acid

print("RNA:", rna_seq)
print("protein:", protein)
```

```python
rna_seq = "AUGGCCAUGACG"
codons = []

for i in range(0, len(rna_seq), 3):
    codon = rna_seq[i:i + 3]

    if len(codon) == 3:
        codons.append(codon)
print(codons)
```

```python
from collections import Counter

codon_counts = Counter(codons)
print(codon_counts)
```

```python
rna_seq = "CCCAUGGCCAAGUAA"
start = rna_seq.find("AUG")
print("start position:", start)
```

```python
stop_codons = {"UAA", "UAG", "UGA"}
```

```python
start = rna_seq.find("AUG")
orf = None

for i in range(start + 3, len(rna_seq) - 2, 3):
    codon = rna_seq[i:i + 3]

    if codon in stop_codons:
        orf = rna_seq[start:i + 3]
        break

print("ORF:", orf)
```
